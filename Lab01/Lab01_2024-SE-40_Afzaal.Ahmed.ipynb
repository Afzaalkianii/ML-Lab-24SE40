Task 8: Check whether "dog" exists in a string

```python
def has_dog(text):
    return "dog" in text.lower()


print(has_dog("I have a dog"))
print(has_dog("I have a cat"))
print(has_dog("I have a DOG"))
```

Task 9: Count how many times "dog" occurs

```python
def count_dog(text):
    return text.lower().count("dog")


print(count_dog("dog cat "))
print(count_dog("cat dog DOG"))
print(count_dog("I have cat and dog"))
```

Task 10:Implement Functions

```python
def caught_speeding(speed, is_birthday):
    allowance = 5 if is_birthday else 0
    adjusted_speed = speed - allowance

    if adjusted_speed <= 60:
        return "No Ticket"
    elif adjusted_speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"
print("Task 10a:", caught_speeding(60, False))  # No Ticket
print("Task 10b:", caught_speeding(70, False))  # Small Ticket
print("Task 10c:", caught_speeding(81, False))  # Big Ticket
print("Task 10d:", caught_speeding(65, True))   # No Ticket
print("Task 10e:", caught_speeding(85, True))   # Small Ticket
print("Task 10f:", caught_speeding(86, True))   # Big Ticket
```

Task 11: Count "dog" with punctuation

```python
import re


def count_dog(text):
    return len(re.findall(r"\bdog\b", text, flags=re.IGNORECASE))


sample_text = "Dog, dog! Dog DOG. dogma is not counted."

print("Dog occurrences:", count_dog(sample_text))
```

Task 12: Check whether a word is a palindrome

```python
def is_palindrome(word):
    return word.lower() == word.lower()[::-1]
print(is_palindrome("level"))   # True
print(is_palindrome("python"))  # False
```

Task 13: Store 5 students and print the topper

```python
students = [
    {"name": "Hassan", "marks": 82},
    {"name": "Usman", "marks": 91},
    {"name": "Fatima", "marks": 87},
    {"name": "Zain", "marks": 95},
    {"name": "Ayesha", "marks": 89}
]
topper = max(students, key=lambda student: student["marks"])
print("Topper:", topper["name"])
print("Marks:", topper["marks"])
```
