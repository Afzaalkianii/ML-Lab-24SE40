**T2(h) answer:** `df[df["Price"] > 1_000_000]` raises a `TypeError` (`'>' not supported between instances of 'str' and 'int'`). The reason is that the `Price` column is stored as **text (strings)**, because some prices were written like `Rs 1,430,000`, which forces pandas to keep the whole column as text. Python cannot say whether a piece of text is "greater than" a number, so the comparison fails. We have to remove `Rs` and commas and convert `Price` to a number first (this is done in Task 3b and 3c).

## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3: Clean the data
# ---- (a) Missing values ----
print("T3a: missing values per column (before)")
print(df.isna().sum())

rows_before = len(df)
df = df.dropna(subset=["Brand", "Model"])
print("T3a: rows removed (no Brand or no Model) =", rows_before - len(df))

df[["Registered City", "Fuel", "Transaction Type"]] = df[["Registered City", "Fuel", "Transaction Type"]].fillna("Unknown")
# Year and KMs Driven are left as NaN on purpose

# ---- (b) Formatting ----
df["Registered City"] = df["Registered City"].str.strip().str.title()
df["Price"] = (df["Price"].str.replace("Rs", "", regex=False)
                          .str.replace(",", "", regex=False)
                          .str.strip())

# ---- (c) Data types ----
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")
print("T3c: dtypes after conversion")
print(df[["Price", "Year"]].dtypes)

# ---- (d) Inconsistent categories ----
print("T3d: different cities BEFORE =", df["Registered City"].nunique())
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
print("T3d: different cities AFTER  =", df["Registered City"].nunique())

# ---- (e) Duplicates ----
print("T3e: exact duplicate ads =", df.duplicated().sum())
df = df.drop_duplicates()
print("T3e: rows after dropping duplicates =", len(df))

# ---- (f) Invalid values ----
bad_kms = (df["KMs Driven"] < 100) | (df["KMs Driven"] > 1_000_000) | df["KMs Driven"].isin([12345, 123456])
print("T3f: KMs Driven values set to NaN =", bad_kms.sum())
df.loc[bad_kms, "KMs Driven"] = np.nan

fake_price = df["Price"].isin([12_345_678, 87_654_321])
too_dear = df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5_000_000)
print("T3f: ads dropped (fake price) =", fake_price.sum(), "| (Suzuki/Daihatsu > 5M) =", too_dear.sum())
df = df[~(fake_price | too_dear)]

# ---- (g) Self-check ----
print("T3g: final shape =", df.shape)
assert df.shape == (20484, 9), "shape does not match the expected (20484, 9)"
print("T3g: shape matches the expected (20484, 9)")
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# T4: Analyse
# (a) price in lakh rupees
df["Price_lakh"] = df["Price"] / 100_000
print("T4a:")
print(df[["Price", "Price_lakh"]].head())

# (b) age of the car in 2020
df["Age"] = 2020 - df["Year"]
print("T4b:")
print(df[["Year", "Age"]].head())

# (c) 5 most common brands
print("T4c: 5 most common brands")
print(df["Brand"].value_counts().head(5))

# (d) 5 cities with the most ads ("Unknown" is not a real city, so it is left out)
print("T4d: 5 cities with the most ads")
print(df.loc[df["Registered City"] != "Unknown", "Registered City"].value_counts().head(5))

# (e) median price per fuel type
print("T4e: median Price_lakh per Fuel (highest first)")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))

# (f) Corolla GLI median price by year, 2010-2018
gli = df[(df["Model"] == "Corolla GLI") & (df["Year"].between(2010, 2018))]
gli_by_year = gli.groupby("Year")["Price_lakh"].median()
print("T4f: Corolla GLI median Price_lakh by year")
print(gli_by_year)
loss_per_year = (gli_by_year.iloc[-1] - gli_by_year.iloc[0]) / (gli_by_year.index[-1] - gli_by_year.index[0])
print("T4f: a Corolla GLI loses about", round(float(loss_per_year), 2), "lakh in median price for every extra year of age")

# (g) 5 most expensive ads
print("T4g: 5 most expensive ads")
print(df.sort_values("Price", ascending=False)[["Brand", "Model", "Year", "KMs Driven", "Price_lakh"]].head(5))

# (h) Cash vs Installment/Leasing
print("T4h: ads and median Price_lakh by Transaction Type")
print(df[df["Transaction Type"].isin(["Cash", "Installment/Leasing"])]
        .groupby("Transaction Type")["Price_lakh"].agg(["count", "median"]))
```

**T4(f) answer:** the median price of a Corolla GLI rises steadily from about 12.3 lakh (2010) to 23.2 lakh (2018), so each extra year of age costs roughly **1.4 lakh** of value (roughly 6% to 12% of the price per year, faster for newer cars).

**T4(g) answer:** yes, several of the top-5 look wrong. The **Toyota Surf 2007 at 410 lakh** and the **Toyota Prius 2015 at 357.5 lakh** are about 20 times the typical price of those models in this data (median about 19.9 and 18.2 lakh), so they are almost certainly typing mistakes (extra zeros). The Land Cruiser ads are also far above the Land Cruiser median (24 lakh), and their odd mileage values (22222, 32222, 3222) look made-up, so they should be treated with caution too.

**T4(h) answer:** Cash ads are far more common (19,634 vs 850) and their median price is **6.9 lakh vs 3.0 lakh** for Installment/Leasing. In this data, installment/leasing ads are mostly for cheaper cars, and they are rare, so cash is the normal way to buy and sell a used car.
