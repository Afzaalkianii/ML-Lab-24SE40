## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# a) Add 5 grace marks
new_marks = np.clip(marks + 5, 0, 100)
print("T9a:", new_marks)

# b) Pass percentage
pass_percentage = np.mean(new_marks >= 50) * 100
print("T9b:", pass_percentage)

# c) Students passed only because of grace marks
index = np.where((marks < 50) & (new_marks >= 50))
print("T9c:", index)

# d) Z-score
z = (new_marks - np.mean(new_marks)) / np.std(new_marks)
print("T9d:", z)
```
